```python
import matplotlib.pyplot as plt
import numpy as np


diffs = {
    'Baseline'   : {'train': 0.0,   'val': 0.0,   'test': 0.0},
    'No vmass'   : {'train': -1.1,  'val': -1.5,  'test': -0.8},
    'No ntracks' : {'train': -1.2,  'val': -0.5,  'test': -0.9},
    'No both'    : {'train': -4.5,  'val': -5.4,  'test': -4.4},
}

# --------------------------------------------------
# 2) Plot
# --------------------------------------------------
groups   = ['Baseline', 'No vmass', 'No ntracks', 'No both']
sets     = ['train', 'val', 'test']
colors = ['#4E79A7',   # train  
          '#59A14F',   # val    
          '#F28E2B']   # test   
x = np.arange(len(groups))                            
bar_w = 0.25                                         

plt.figure(figsize=(8, 5))

for i, s in enumerate(sets):
    heights = [diffs[g][s] for g in groups]
    pos = x + (i - 1) * bar_w                        
    bars = plt.bar(pos, heights, width=bar_w,
                   color=colors[i], label=s.capitalize())


    for bar, h in zip(bars, heights):
        plt.text(bar.get_x() + bar.get_width() / 2,
                 h - 0.15,                           
                 f'{h:.1f}',
                 ha='center', va='top', fontsize=12)

plt.axhline(0, color='grey', linewidth=0.8)
plt.ylim(-6, 0)
plt.ylabel('Accuracy difference to baseline (pp)')
plt.title('Performance drop after removing vertex-scale features')
plt.xticks(x, groups)
plt.legend(title='Dataset', frameon=False)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()

```
